```python
N = 8

board = [[0 for _ in range(N)] for _ in range(N)]

def is_safe(board, row, col):

    # Check left side
    for i in range(col):
        if board[row][i] == 1:
            return False

    # Upper diagonal
    i, j = row, col
    while i >= 0 and j >= 0:
        if board[i][j] == 1:
            return False
        i -= 1
        j -= 1

    # Lower diagonal
    i, j = row, col
    while i < N and j >= 0:
        if board[i][j] == 1:
            return False
        i += 1
        j -= 1

    return True

def solve(board, col):

    if col >= N:
        return True

    for row in range(N):

        if is_safe(board, row, col):

            board[row][col] = 1

            if solve(board, col + 1):
                return True

            board[row][col] = 0

    return False

if solve(board, 0):

    print("Solution Found:")
    for row in board:
        print(row)

else:
    print("No Solution Exists")
```

```python
students = ["A", "B", "C"]
seats = [1, 2, 3]
assignment = {}
def is_valid(student, seat):
    if student == "A" and seat == 2:
        return False
    if seat in assignment.values():
        return False
    return True
def backtrack(index):
    if index == len(students):
        return True
    student = students[index]
    for seat in seats:
        if is_valid(student, seat):
            assignment[student] = seat
            if backtrack(index + 1):
                return True
            del assignment[student]
    return False
if backtrack(0):
    print("Solution Found:")
    for student, seat in assignment.items():
        print(f"{student} --> Seat {seat}")
else:
    print("No solution exists.")
```

```python
# -------------------------------
# Backward Chaining Example
# -------------------------------

# Facts stored in knowledge base
facts = {
    "Human(Socrates)"
}

# Rules stored as:
# conclusion : premise
rules = {
    "Mortal(Socrates)": "Human(Socrates)"
}

# Recursive backward chaining function
def backward_chain(goal):

    print(f"\nTrying to prove: {goal}")

    # Step 1: Check whether goal is already a fact
    if goal in facts:
        print(f"{goal} is a FACT.")
        return True

    # Step 2: Check if any rule can prove it
    if goal in rules:
        premise = rules[goal]
        print(f"Need to prove: {premise}")

        if backward_chain(premise):
            print(f"Rule applied:")
            print(f"{premise} --> {goal}")
            return True

    # If the goal was not a fact and no rule (or its premise) could prove it
    print(f"Cannot prove {goal}")
    return False


# Goal
goal = "Mortal(Socrates)"

print("Backward Chaining")

if backward_chain(goal):
    print("\nGoal PROVED")
else:
    print("\nGoal NOT PROVED")
```

```python
# Towers of Hanoi

def hanoi(n, source, auxiliary, destination):
    if n == 1:
        print("Move disk 1 from", source, "to", destination)
        return
    hanoi(n - 1, source, destination, auxiliary)
    print("Move disk", n, "from", source, "to", destination)
    hanoi(n - 1, auxiliary, source, destination)

hanoi(5, "Stand 1", "Stand 2", "Stand 3")
```
