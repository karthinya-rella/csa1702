```python
# Towers of Hanoi
def hanoi(n, source, auxiliary, destination):
    if n == 1:
        print("Move disk 1 from", source, "to", destination)
        return
    hanoi(n - 1, source, destination, auxiliary)
    print("Move disk", n, "from", source, "to", destination)
    hanoi(n - 1, auxiliary, source, destination)

hanoi(5, "Stand 1", "Stand 2", "Stand 3")
```

```python
# Family Tree
father = {
    "raju": "ramesh",
    "sita": "ramesh",
    "arun": "mohan"
}

mother = {
    "raju": "geeta",
    "sita": "geeta",
    "arun": "radha"
}

def parent(child):
    if child in father:
        print(father[child], "is the father of", child)
        print(mother[child], "is the mother of", child)
    else:
        print("No information available")

child = input("Enter child's name: ").lower()
parent(child)
```

```python
# Fruit and Its Color using Backtracking
fruits = {
    "apple": "red",
    "banana": "yellow",
    "orange": "orange",
    "grapes": "green",
    "mango": "yellow"
}
print("Fruit and Color:")
for fruit, color in fruits.items():
    print(fruit, "is", color)
fruit = input("\nEnter fruit name: ").lower()
if fruit in fruits:
    print("Color of", fruit, "is", fruits[fruit])
else:
    print("Fruit not found")
```

```python
# Best First Search
import heapq
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': ['G'],
    'G': []
}
heuristic = {'A': 6, 'B': 4, 'C': 5, 'D': 2, 'E': 3, 'F': 1, 'G': 0}
def best_first_search(start, goal):
    queue = [(heuristic[start], start)]
    visited = set()
    while queue:
        h, node = heapq.heappop(queue)
        if node in visited:
            continue
        visited.add(node)
        print(node, end=" ")
        if node == goal:
            print("\nGoal found")
            return
        for neighbor in graph[node]:
            if neighbor not in visited:
                heapq.heappush(queue, (heuristic[neighbor], neighbor))
best_first_search('A', 'G')
```

```python
# Medical Diagnosis
symptoms = {
    "fever": False,
    "cough": False,
    "headache": False,
    "body_pain": False,
    "sore_throat": False
}
for symptom in symptoms:
    answer = input("Does the patient have " + symptom + "? (yes/no): ").lower()
    symptoms[symptom] = answer == "yes"
if symptoms["fever"] and symptoms["cough"] and symptoms["body_pain"]:
    print("Diagnosis: Flu")
elif symptoms["fever"] and symptoms["cough"] and symptoms["sore_throat"]:
    print("Diagnosis: Common Cold")
elif symptoms["headache"] and symptoms["fever"]:
    print("Diagnosis: Fever")
else:
    print("Diagnosis: Unknown")
```

```python
# Forward Chaining
facts = {"A", "B"}
rules = [
    ({"A", "B"}, "C"),
    ({"C"}, "D"),
    ({"D"}, "E")
]
print("Initial Facts:", facts)
changed = True
while changed:
    changed = False
    for conditions, conclusion in rules:
        if conditions.issubset(facts) and conclusion not in facts:
            facts.add(conclusion)
            print("Derived:", conclusion)
            changed = True
print("Final Facts:", facts)
```
